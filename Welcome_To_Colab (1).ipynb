```
import heapq


def heuristic(a, b):
    return abs(a[0] - b[0]) + abs(a[1] - b[1])

def astar(grid, start, goal):

    q = [(heuristic(start, goal), 0, start, [start])]
    cost = {start: 0}

    while q:

        f, g, current, path = heapq.heappop(q);

        if current == goal:
            return path, g


        for dx, dy in [(1, 0), (-1, 0), (0, 1), (0, -1)]:
            neighbor = (current[0] + dx, current[1] + dy)


            if 0 <= neighbor[0] < len(grid) and 0 <= neighbor[1] < len(grid[0]):

                if grid[neighbor[0]][neighbor[1]] == 0:
                    new_cost = g + 1
                    if neighbor not in cost or new_cost < cost[neighbor]:
                        cost[neighbor] = new_cost
                        f_score = new_cost + heuristic(neighbor, goal)
                        heapq.heappush(q, (f_score, new_cost, neighbor, path + [neighbor]))

    return None, float('inf')


grid = [
    [0, 0, 0],
    [0, 1, 0],
    [0, 0, 0]
]

start = (0, 0)
goal = (2, 2)

print("Input grid:")
for row in grid:
    print(row)
print("Start:", start);
print("Goal:", goal)

optimal_path, total_cost = astar(grid, start, goal)

if optimal_path:
    print("Optimal path:", optimal_path)
    print("Cost:", total_cost)
else:
    print("No path found.")
```
